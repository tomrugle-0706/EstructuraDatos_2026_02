```python
def potencia_dos_recursion1(n):
  if n==0:
    return 1
  else:
    a=potencia_dos_recursion(n-1)
    return 2*a
```

```python
def potencia_dos_recursion2(n):
  if n==0:
    return 1
  potencia= potencia_dos_recursion1(n//2)
  if n%2==0:
    return potencia*potencia
  else:
    return 2*potencia*potencia
```

```python
import os

def disk_usage(path):
  """Return the number of bytes used by a file/folder and any descendents."""
  total = 0
  try:
    total = os.path.getsize(path)                  # account for direct usage
  except FileNotFoundError:
    print(f"Warning: File or directory not found at '{path}'. Skipping this item.")
    return 0
  except OSError as e:
    print(f"Warning: Error accessing '{path}': {e}. Skipping this item.")
    return 0

  if os.path.isdir(path):                        # if this is a directory,
    try:
      for filename in os.listdir(path):            # then for each child:
        childpath = os.path.join(path, filename)   # compose full path to child
        total += disk_usage(childpath)             # add child's usage to total
    except FileNotFoundError:
        print(f"Warning: Directory contents not found for '{path}'. Skipping contents.")
    except OSError as e:
        print(f"Warning: Error listing directory '{path}': {e}. Skipping contents.")

  # Use the convert_bytes function for descriptive output
  print ('{0:<7}'.format(convert_bytes(total)), path)           # descriptive output (optional)
  return total                                   # return entire usage
```

```python
def convert_bytes(num_bytes):
  """Convert bytes to human-readable format (e.g., KB, MB, GB)."""
  for unit in ['bytes', 'KB', 'MB', 'GB', 'TB']:
    if num_bytes < 1024.0:
      return f"{num_bytes:.2f} {unit}"
    num_bytes /= 1024.0
  return f"{num_bytes:.2f} PB" # For extremely large numbers

disk_usage('/content/drive')
```

```python
def maximo_recursivo(L):
  if len(L)==1:
    return L[0]
  maximotentativo= maximo_recursivo(L[:-1])
  if maximotentativo>L[-1]:
    return maximotentativo
  else:
    return L[-1]
```

```python
maximo_recursivo([1,2,4,5,6,7,-10,14,23])
```

```python
def maximo_recursivocuenta(L):
  if len(L)==1:
    return L[0]
  count=1
  maximotentativo,count= maximo_recursivocuenta(L[:-1]),1
  if maximotentativo>L[-1]:
    return maximotentativo
  elif maximotentativo==L[-1]:
      return L, count+1
  else:
    return L[-1],count
```
